### Listas

1. Soma dos Elementos

```python
# Dada uma lista de números, calcule a soma de todos os elementos
numeros = [5, 10, 15, 20, 25]
soma = sum(numeros)

# Seu código aqui

print(f"Soma: {soma}")  # Deve imprimir 75
```

2. Encontrar o Maior Número

```python
# Encontre o maior número em uma lista
numeros = [34, 12, 89, 5, 67, 23]
maior = max(numeros)

# Seu código aqui

print(f"Maior número: {maior}")  # Deve imprimir 89
```

3. Adicionar Elementos com Condição

```python
# Dada uma lista de números, crie uma nova lista apenas com números maiores que 50
numeros = [25, 60, 15, 80, 45, 90, 30]
maiores_que_50 = []
quant = len(numeros)
# Seu código aqui (use append com if)
for i in range(quant):
    if numeros[i] >= 50:
        maiores_que_50.append(numeros[i])
    else:
        print(f"O número {numeros[i]} não é maior que 50.")

print(f"\n\n{maiores_que_50}")  # Deve imprimir: [60, 80, 90]
```

4.  Criar Lista de Vogais de uma Frase

```python
# Crie uma lista com todas as vogais de uma frase
frase = "Python é uma linguagem de programação"
vogais = []

# Seu código aqui (use append com if)
vogais_referencia = "aeiouáéíóúâêîôûãõAEIOUÁÉÍÓÚÂÊÎÔÛÃÕ"

for letra in frase:
    if letra in vogais_referencia:
        vogais.append(letra)

print(vogais)  # Deve imprimir: ['o', 'é', 'u', 'a', 'i', 'u', 'a', 'e', 'o', 'a', 'a', 'o']
```

5. Controle de Gastos Pessoais

```python
# Programa de Controle de Gastos
print("=== CONTROLE DE GASTOS PESSOAIS ===")

# Listas para armazenar os dados
nomes_gastos = []
valores_gastos = []

# Pergunta a quantidade de gastos
quant_gastos = int(input("Qual a sua quantidade de gastos ? Ex: 10"))

# Loop para coletar os gastos
for gastos in range(quant_gastos):
    nome_gasto = input(f"Digite o nome do seu {gastos + 1}º gasto")
    valor_gasto = float(input(f"Digite o valor do seu {gastos + 1}º gasto"))
# Adiciona às listas usando APPEND
    nomes_gastos.append(nome_gasto)
    valores_gastos.append(valor_gasto)

# Calcula o total dos gastos
total_gastos = sum(valores_gastos)

# Exibe os resultados
print(f"TOTAL GASTO: R$ {total_gastos:.2f}")

# Verifica se está gastando muito
LIMITE_GASTO = 1650
economia = LIMITE_GASTO - total_gastos
excesso = total_gastos - LIMITE_GASTO

if total_gastos <= LIMITE_GASTO:
    print(f"\nVocê gastou R$:{total_gastos}, economizando R$:{economia} e está dentro do limite aceitável de gastos, continue assim!!!")

elif total_gastos <= LIMITE_GASTO*2:
    print(f"\nVocê gastou R$:{total_gastos}, sendo R$:{excesso} a mais do que deveria, tome cuidade!!!")

else:
    print(f"\nVocê gastou R$:{total_gastos} e estorou o limite do que deveria ter gastado, sendo R$:{excesso} a mais do que deveria, reveja suas atitudes imediatamente!!!")
```
